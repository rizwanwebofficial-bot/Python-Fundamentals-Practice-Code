```python
marks = 67
passing_marks = 50
attendance = 78
required_attendance = 75

# Write a program that checks and prints:
    
if marks>=passing_marks:
    print("student is passed.")
    
else:
    print("student is fail.")
    
if attendance >= required_attendance:
    print("enough attendence")
    
else:
    print("no enugh attendence")
    

if marks>=passing_marks and attendance >= required_attendance:
    print("both conditons met")
else:
    print("both conditions not met")
    
if marks>80:
    print("greater than 80.")
    
else:
    print("less than 80.")
    
if marks==50:
    print("equal to 50.")
    
else:
    print("not equal to 50.")
```

Loops

```python
# Write a program that takes a number n from the user and:

# Uses a for loop to print all numbers from 1 to n.
# Uses a while loop to calculate the sum of all numbers from 1 to n.
# Print how many numbers between 1 and n are even and how many are odd.

n = int(input("Enter a number:"))


for i in range(1, n + 1):
    print(i)
    
i=1
total = 0
    
while (i<=n ):
    total = total + i
    i = i + 1

print(f"Sum of numbers from 1 to {n} is: {total}")

for i in range(1, n + 1):
    if i % 2 == 0:
        print(f"{i} is even")
    else:
        print(f"{i} is odd")
```
